### **Data Cleaning**

```python
print(blink_data["Item Fat Content"].unique())
```

```python
#there is a duplicate so we should clean it
blink_data["Item Fat Content"]=blink_data["Item Fat Content"].replace({'LF':'Low Fat',
                                                                      'low fat':'Low Fat',
                                                                      'reg':'Regular'})
```

```python
print(blink_data["Item Fat Content"].unique())
```

### **BUSINESS REQUIREMENT**

#### **KPI's REQUIREMENTS**

```python
#total sales
tot_sales=blink_data['Sales'].sum()

#avg sales
avg_sales=blink_data[Item Fat Content'].mean()

#no of item sold
no_of_item=blink_data['Sales'].count()

#avg rating
avg_rating=blink_data['Rating'].mean()

print(f'total sales : ${tot_sales:,.2f}')
print(f'Average sales : ${avg_sales:,.2f}')
print(f'No of item : {no_of_item}')
print(f'Average rating : {avg_rating:,.2f}')
```

### **CHART REQUIREMENTS**

#### **1. Total Sales by Fat Content**

```python
blink_data["Item Fat Content"].unique()
```

```python
sales_by_fat=blink_data.groupby("Item Fat Content")['Sales'].sum()
sales_by_fat
```

```python
plt.pie(sales_by_fat,labels=sales_by_fat.index,autopct='%.1f%%',
       startangle=90)
plt.title("Sales by Fat")
plt.show()
```

### **2. Total Sales by Item Type**

```python
tot_sal_by_item_type=blink_data.groupby('Item Type')['Sales'].sum().sort_values(ascending=False)
plt.figure(figsize=(10,6))
bars=plt.bar(tot_sal_by_item_type.index,tot_sal_by_item_type.values)
plt.xticks(rotation=-90)
plt.xlabel("Item type")
plt.ylabel("Total sales")

for bar in bars:
    plt.text(bar.get_x() + bar.get_width() / 2,bar.get_height(),
            f'{bar.get_height():,.0f}',ha='center',va='bottom',fontsize=8)
plt.tight_layout()
plt.show()
```

### **3. Fat Content by Outlet for Total Sales**

```python
grouped=blink_data.groupby(['Outlet Location Type','Item Fat Content'])['Sales'].sum().unstack()
grouped
```

```python
grouped=grouped[['Regular','Low Fat']]
grouped
```

```python
ax=grouped.plot(kind='bar',figsize=(8,5),title='Outlet Tier by Item Fat Content')
plt.xlabel('outlet location tier')
plt.ylabel('Total Sales')
plt.legend(title='Item fat Content')
plt.tight_layout()
plt.show()
```

### **4. Total Sales by Outlet Establishment**

```python
sales_by_year=blink_data.groupby('Outlet Establishment Year')['Sales'].sum().sort_index()
plt.figure(figsize=(9,5))
plt.plot(sales_by_year.index,sales_by_year.values,marker='o',linestyle='-')

plt.xlabel('Outlet Establishment Year')
plt.ylabel('Total Sales')
plt.title('Outlet Establishment')

for x,y in zip(sales_by_year.index,sales_by_year.values):
    plt.text(x,y,f'{y:,.0f}',ha='center',va='bottom',fontsize=8)

plt.tight_layout()
plt.show()
```

### **Sales by outlet size**

```python
sales_by_size=blink_data.groupby('Outlet Size')['Sales'].sum()
sales_by_size
```

```python
plt.pie(sales_by_size,labels=sales_by_size.index,autopct='%1.1f%%',startangle=90)
plt.title('Outlet Size')
plt.tight_layout()
plt.show()
```

### **Sales by Outlet Location**

```python
sales_by_location=blink_data.groupby('Outlet Location Type')['Sales'].sum().reset_index()
sales_by_location=sales_by_location.sort_values('Sales',ascending=False)

plt.figure(figsize=(8,3))
ax=sns.barplot(x='Sales',y='Outlet Location Type',data=sales_by_location)

for y,x in zip(sales_by_location['Outlet Location Type'].values,sales_by_location['Sales'].values):
    plt.text(x,y,f"{x:,.0f}",ha='left',va='bottom',fontsize=8)

plt.title('Total Sales by Outlet location Type')
plt.xlabel('Total Sales')
plt.ylabel('Outlet Location Type')

plt.tight_layout()
plt.show()
```
